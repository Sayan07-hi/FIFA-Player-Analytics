## Investigation of missing club names and league names
I found that 89 players have missing values in both the `club_name` and `league_name` columns.

Initially, I assumed these players were Icons or Legends. However, after inspecting the records, I found a mixture of player types:

- Free agents or unattached players
- Players whose transfers may have been in progress when the dataset was created
- A few active professional players with missing club/league information

Therefore, the missing values cannot be explained by a single reason. They are likely caused by a combination of player status and limitations of the dataset snapshot.

```python
print(df[df["league_name"].isna()]["short_name"])
```

```python
missing_players = df[df["league_name"].isnull()]
missing_players[["long_name", "club_name", "league_name"]]
```

## Missing Value Handling Decisions 
- Dropped columns with excessive missing values (>90%)
- Kept columns regarding player stats where missing values are expected for player roles

```python
missing_values_percent = (df.isna().sum() / len(df)) * 100
missing_values_percent = missing_values_percent[missing_values_percent > 0]
print(missing_values_percent)
```

```python
df.drop(columns=["work_rate","nation_team_id","nation_position","nation_jersey_number","player_tags","club_loaned_from"], inplace=True)
```

```python
missing_values_percent = (df.isna().sum() / len(df)) * 100
missing_values_percent = missing_values_percent[missing_values_percent > 0]
print(missing_values_percent)
```
