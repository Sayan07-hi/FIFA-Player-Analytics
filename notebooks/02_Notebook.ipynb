# Exploratory Data Analysis (EDA)

## Objective

- The goal of this notebook is to understand the cleaned FC26 player dataset by exploring player attributes, distributions, relationships, and trends before building machine learning models.

- The insights gained from this analysis will help identify important features that influence a player's market value.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("C:\\Users\\SAYANJIT\\OneDrive\\Desktop\\project\\FIFA Player Analytics & Market Value Prediction\\data\\processed\\fc26_clean.csv")
df
```

```python
df.describe()
```

```python
#Market value graph
plt.figure(figsize=(15,5))
plt.subplot(1,3,1)
plt.hist(df["value_eur"],bins=10,edgecolor="black")
plt.title("Player Market Value")
plt.xlabel("Market Value (EUR)")
plt.ylabel("Frequency")

#Age distribution graph
plt.subplot(1,3,2)
plt.hist(df["age"],bins=10,edgecolor="black")
plt.title("Player Age")
plt.xlabel("Age")
plt.ylabel("Frequency")

#Overall rating distribution graph
plt.subplot(1,3,3)
plt.hist(df["overall"],bins=10,edgecolor="black")
plt.title("Player Overall Rating")
plt.xlabel("Overall Rating")
plt.ylabel("Frequency")
plt.tight_layout(pad=1)

plt.show()
```

- *Market Value Distribution* is heavily right skewed due to lower market values of most players while small number of players have extremely high values. Because of these values the x-axis is compressed making it diofficult to visualize lower value range clearly.
- *Ages* of most players are approximately between 20 and 30 years old, with very few players at extreme ages.
- *Overall Rating* have most players are centered around the mid-60s, while high rated players are comparatively rare.

```python
#Graph for top 10 nationalities
plt.figure(figsize=(18, 5))
plt.subplot(1,3,1)
top10nationality = df["nationality_name"].value_counts().head(10)
plt.bar(top10nationality.index, top10nationality.values)
plt.xlabel("Nationality")
plt.ylabel("Number of Players")
plt.title("Top 10 Nationalities")
plt.xticks(rotation=45)

#Graph for top 10 Clubs by Number of Players
top10clubs = df["club_name"].value_counts().head(10)
plt.subplot(1,3,2)
plt.bar(top10clubs.index, top10clubs.values)
plt.xlabel("Club")
plt.ylabel("Number of Players")
plt.title("Top 10 Clubs")
plt.xticks(rotation=60, ha="right")

#Graph for top 10 positions
plt.subplot(1,3,3)
df["first_pos"] = df["player_positions"].str.split(",").str[0]
unique_positions = df["first_pos"].value_counts().head(10)
plt.bar(unique_positions.index, unique_positions.values)
plt.xlabel("Position")
plt.ylabel("Number of Players")
plt.title("Top 10 Positions")
plt.xticks(rotation=45)

plt.tight_layout(pad=1)
plt.show()
```

- England has the highest number of players in the dataset, followed by Germany and Argentina. European countries dominate the dataset, with England, Germany, Spain, France, Italy, the Netherlands, and Norway appearing in the top 10. South American football is strongly represented by Argentina and Brazil, while the United States is the only North American country in the top.
- The clubs with the highest number of player records are shown. This visualization represents that some clubs have larger registered squads or more complete player records in the dataset.
- Centre Backs (CB) are the most common primary position in the dataset. Strikers (ST) and Central Midfielders (CM) are also among the most frequently occurring positions. Goalkeepers (GK), Fullbacks (RB, LB), and Wide Midfielders (LM, RM) appear less frequently compared to central positions.
